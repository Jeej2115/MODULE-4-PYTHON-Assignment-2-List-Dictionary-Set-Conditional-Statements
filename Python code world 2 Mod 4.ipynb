```python
# Initial lists
age_list = [24, 25, 27, 28, 29]
name_list = ["John", "Sam", "Rahul", "Anita", "Kumar"]

# a. Append "Yazhini" to name_list
name_list.append("Yazhini")
print("After appending Yazhini:", name_list)

# b. Insert 26 at index 2 in age_list
age_list.insert(2, 26)
print("After inserting 26:", age_list)

# c. Remove "Yazhini" from name_list
name_list.remove("Yazhini")
print("After removing Yazhini:", name_list)

# d. Pop the last element from age_list
removed_age = age_list.pop()
print("Removed age:", removed_age)
print("After popping the last element:", age_list)

# e. Extend age_list with [31, 30, 32]
age_list.extend([31, 30, 32])
print("After extending age_list:", age_list)

# f. Sort age_list in descending order
age_list.sort(reverse=True)
print("Age list in descending order:", age_list)

# g. Find maximum age, minimum age, and sum of all ages
max_age = max(age_list)
min_age = min(age_list)
total_age = sum(age_list)

print("Maximum age:", max_age)
print("Minimum age:", min_age)
print("Sum of all ages:", total_age)

```

```python
# Accessing List Elements

# a. Print the first element of name_list
print("First element:", name_list[0])

# b. Print the last element of name_list
print("Last element:", name_list[-1])

# c. Print the elements from index 2 to index 4
print("Elements from index 2 to index 4:", name_list[2:5])

# d. Print the elements of name_list in reverse order
print("Name list in reverse order:", name_list[::-1])
```

```python
# Dictionary (Creation, Modification and Access)

# a. Create a dictionary with five students and their marks
student_marks = {
    "John": 65,
    "Sam": 76,
    "Rahul": 92,
    "Anita": 78,
    "Kumar": 95
}

print("Student Marks:", student_marks)

# b. Access and print the mark of a specific student
print("Rahul's mark:", student_marks["Rahul"])

# c. Add a new student "Janani" with a mark of 80
student_marks["Janani"] = 80
print("After adding Janani:", student_marks)

# d. Update the mark of an existing student to 82
student_marks["Anita"] = 82
print("After updating Anita's mark:", student_marks)

# e. Print all keys, values, and key-value pairs

print("All student names (Keys):", student_marks.keys())
print("All marks (Values):", student_marks.values())
print("All key-value pairs (Items):", student_marks.items())
```

```python
# Sets – Operations

# Create a set with vowels, including duplicate values
my_set = {'a', 'e', 'i', 'o', 'u', 'a', 'a', 'i'}

# Print the set
print("My Set:", my_set)
```

```python
# Attempt to access my_set[4]

print(my_set[4])
```

```python
# Explanation:Sets do not support indexing because they are unordered collections. 
#Therefore, we cannot access an element using an index such as my_set[4].
```

```python
# Create Two Sets

# Create the first set
set1 = {1, 3, 5, 7, 9}

# Create the second set
set2 = {2, 3, 5, 8, 10}

print("Set 1:", set1)
print("Set 2:", set2)
```

```python
# Find Union and Intersection

# Compute the union of set1 and set2
union_set = set1.union(set2)

# Compute the intersection of set1 and set2
intersection_set = set1.intersection(set2)

# Display the results
print("Union of set1 and set2:", union_set)
print("Intersection of set1 and set2:", intersection_set)
```

```python
# Find the Performance Category

# Use if, elif, and else statements to categorize the user's score.

# Prompt the user to enter a score
score = int(input("Enter your score (0 to 10): "))

# Check the performance category
if score > 7:
    print("Performance Category: Above Average")
    print("Excellent performance! Keep up the good work.")

elif score >= 4 and score <= 7:
    print("Performance Category: Average")
    print("Good effort! With more practice, you can improve further.")

else:
    print("Performance Category: Below Average")
    print("Need to improve your performance; consistent practice will lead to better results.")

```
